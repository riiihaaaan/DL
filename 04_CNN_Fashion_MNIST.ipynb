# Practical Assignment 4: Image Classification using CNN on Fashion MNIST Dataset

**Aim:** Build an image classification model using Convolutional Neural Networks (CNN) on the Fashion MNIST clothing dataset.

**Key Concepts to Remember:**
1. **Input Shape:** 2D convolution requires spatial dimensions: `(28, 28, 1)`.
2. **Feature Extraction:** `Conv2D` learns spatial feature maps; `MaxPooling2D` reduces spatial dimensions.
3. **Classifier Head:** `Flatten` $\to$ `Dense` $\to$ `Softmax` output.


```python
import os
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'
import pandas as pd
import numpy as np
import tensorflow as tf
from sklearn.metrics import classification_report

# Step 1: Load and Reshape Data
df_train = pd.read_csv(r'd:/workspace01/4th year/DL/LabPracticesIV_DEEP_LEARNING/dataset/Fashion MNIST/fashion-mnist_train.csv', nrows=6000)
df_test = pd.read_csv(r'd:/workspace01/4th year/DL/LabPracticesIV_DEEP_LEARNING/dataset/Fashion MNIST/fashion-mnist_test.csv', nrows=1500)

y_train = df_train.iloc[:, 0].values.astype('int32')
X_train = df_train.iloc[:, 1:].values.astype('float32').reshape(-1, 28, 28, 1) / 255.0
y_test = df_test.iloc[:, 0].values.astype('int32')
X_test = df_test.iloc[:, 1:].values.astype('float32').reshape(-1, 28, 28, 1) / 255.0

print(f'X_train shape: {X_train.shape}, X_test shape: {X_test.shape}')

# Step 2: Build CNN Architecture
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(28, 28, 1)),
    tf.keras.layers.Conv2D(32, (3, 3), activation='relu'),
    tf.keras.layers.MaxPooling2D((2, 2)),
    tf.keras.layers.Conv2D(64, (3, 3), activation='relu'),
    tf.keras.layers.MaxPooling2D((2, 2)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dropout(0.25),
    tf.keras.layers.Dense(10, activation='softmax')
])
model.summary()

# Step 3: Compile and Train Model
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
history = model.fit(X_train, y_train, epochs=4, batch_size=64, validation_split=0.1, verbose=1)

# Step 4: Evaluate Model
loss, acc = model.evaluate(X_test, y_test, verbose=0)
print(f'\nFashion MNIST Test Accuracy: {acc*100:.2f}% | Test Loss: {loss:.4f}')

class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat', 'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']
y_pred = np.argmax(model.predict(X_test, verbose=0), axis=1)
print('\nClassification Report:')
print(classification_report(y_test, y_pred, target_names=class_names, digits=4))
```
